## Przykład obliczeń

Dla ciągu geometrycznego o pierwszym wyrazie $a_1$ i ilorazie $q\neq 1$ suma pierwszych $n$ wyrazów wynosi

$$
S_n=a_1\frac{1-q^n}{1-q}.
$$

Poniższe komórki pokazują proste obliczenia, zapis wyników do plików, tworzenie wykresów oraz generowanie animacji.

```python
import numpy as np
import matplotlib.pyplot as plt

k = np.arange(1, 21)
plt.figure(figsize=(6, 4))
plt.plot(k, k**2, marker="o")
plt.xlabel("k")
plt.ylabel(r"$k^2$")
plt.title(r"Wykres $y=k^2$")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("wykres.png", dpi=150)
plt.show()
```

```python
# W Colabie ffmpeg jest standardowo dostępny.
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import Video, display

fig, ax = plt.subplots(figsize=(5, 4))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect("equal")
ax.axis("off")
line, = ax.plot([], [], "k-", lw=2)
bob, = ax.plot([], [], "o", ms=12)

def update(frame):
    theta = 0.65*np.cos(frame)
    x = np.sin(theta)
    y = -np.cos(theta)
    line.set_data([0, x], [0, y])
    bob.set_data([x], [y])
    return line, bob

frames = np.linspace(0, 4*np.pi, 120)
anim = FuncAnimation(fig, update, frames=frames, blit=True)
anim.save("video.mp4", writer=FFMpegWriter(fps=30))
plt.close(fig)
display(Video("video.mp4", embed=True))
```

```python
# Konwersja MP4 -> GIF przy pomocy ffmpeg.
import subprocess
subprocess.run([
    "ffmpeg", "-y", "-i", "video.mp4",
    "-vf", "fps=15,scale=480:-1:flags=lanczos",
    "video.gif"
], check=True)
print("Zapisano video.gif")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Inna krzywa parametryczna: torus knot-like curve
t = np.linspace(0, 6*np.pi, 600)
x = (1 + 0.35*np.cos(5*t))*np.cos(t)
y = (1 + 0.35*np.cos(5*t))*np.sin(t)
z = 0.35*np.sin(5*t)

fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection="3d")
ax.plot(x, y, z, label="Modified parametric curve")
ax.set_title("3D Parametric Curve")
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()
plt.tight_layout()
plt.savefig("parametric_curve.png", dpi=150)
plt.show()
```
